```python
df.groupby('Country').agg({
    'Age':['max','min'],
    'Watch_Time_Hours':'sum'
})
```

```python
country_agg = df.groupby('Country').agg(
    Max_Age=('Age','max'),
    Min_Age=('Age','min'),
    Total_Watch_Time=('Watch_Time_Hours','sum')
)
```
