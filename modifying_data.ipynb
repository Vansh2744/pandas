```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('netflix_users.csv')
df
```

```python
df['Username'] = df['Name'].str.lower().str.split(" ").str.join("") + "@gmail.com"
```

```python
df.head()
```

```python
def check_watch(hrs):
    if hrs > 100:
        return "Binge Watching"
    else:
        return "Normal Watching"
```

```python
df['Watching Type'] = df['Watch_Time_Hours'].apply(check_watch)
```

```python
df.head()
```

```python
subs_codes = {'Premium':'PRM', 'Basic':'BSC', 'Standard':'STD'}
df['Subscription Codes']=df['Subscription_Type'].map(subs_codes)
```
