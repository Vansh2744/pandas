```python
df['Watch_Rank'] = df['Watch_Time_Hours'].rank()
```

```python
df.sort_values(by='Watch_Rank')
```

```python
df['Watch_Rank'] = df['Watch_Time_Hours'].rank(method="min")
```

```python
df.sort_values(by='Watch_Rank')
```

```python
df['Watch_Rank'] = df['Watch_Time_Hours'].rank(method="max")
```

```python
df.sort_values(by='Watch_Rank')
```

```python
df['Watch_Rank'] = df['Watch_Time_Hours'].rank(method="dense")
```

```python
df.sort_values(by='Watch_Rank')
```
